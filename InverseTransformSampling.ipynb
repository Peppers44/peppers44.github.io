```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import expon
#please set your seed to this for repeatable results
np.random.seed(67)
```

The premise of inverse transform sampling is to take a uniform distribution from (0,1), pick random numbers and plug those into the inverse CDF of a chosen continuous distribution. It works because the inverse CDF spits out the value that percentile is related to. 

For an Exponential distribution with mean (I chose Exponential because I got really familiar during Exam P prep) $\theta$:
1. CDF: $F(t) = 1 - e^{-t/\theta} = U$
2. Solve for t: $e^{-t/\theta} = 1 - U$
3. Inverse CDF: $t = -\theta \ln(1 - U)$



```python
def inverseTransform(samples, theta):
    #We are generating our random uniform samples here.
    U = np.random.uniform(0, 1, samples)
    
    #Inverse CDF
    T = -theta * np.log(U)
    
    return T

#We are simulating 10000 claims with a mean of 6000
mean = 6000
simulatedclaims = inverseTransform(10000, mean)
```

```python
plt.figure(figsize=(10, 6))
plt.hist(simulated_claims, bins=50, density=True, alpha=0.6, 
         color='skyblue', edgecolor='black', label='Simulated Data')
```

And there we have it, inverse transform sampling.
